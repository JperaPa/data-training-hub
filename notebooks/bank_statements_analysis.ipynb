```python
import pandas as pd
import glob

# Load all CSVs inside data/raw/bank_statements/
path = "../data/raw/bank_statements/*.csv"

files = glob.glob(path)

print("Files found:", files)

dfs = [pd.read_csv(f) for f in files]

# Combine all CSVs into one dataframe
df = pd.concat(dfs, ignore_index=True)

df.head()
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("-", "_")
)

df.head()
```

```python
if "date" in df.columns:
    df["date"] = pd.to_datetime(df["date"], errors="coerce")

df.head()
```

```python
def categorize(description):
    description = str(description).lower()

    if "amazon" in description:
        return "Shopping"
    if "walmart" in description or "jewel" in description:
        return "Groceries"
    if "shell" in description or "bp" in description:
        return "Gas"
    if "mortgage" in description:
        return "Housing"
    if "t-mobile" in description:
        return "Phone"
    if "transfer" in description:
        return "Transfer"
    return "Other"

if "description" in df.columns:
    df["category"] = df["description"].apply(categorize)

df.head()
```

```python
df.to_csv("../data/processed/bank_transactions_clean.csv", index=False)

df.tail()
```
